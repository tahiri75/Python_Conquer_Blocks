1. Escribe un programa en Python para encontrar los elementos duplicados de una lista,
añadirlos a una nueva lista y borrarlos de la lista. Después imprime una lista con tan solo los
elementos únicos.

```python
# --- crear la lista original ---
originalList = []
listLen = int(input("Teclee la longitud de la lista: "))
for i in range(0, listLen):
    originalList.append(input("Elemento {} de la lista: ".format(i)))

# --- mostrar la lista original ---
print("Lista original:", originalList)

# --- buscar duplicados
duplicateList = []
i = 0

while i < len(originalList):
        j = i + 1
        while j < len(originalList):
            if (originalList[i] == originalList[j]) and not (
                originalList[i] in duplicateList
            ):
                # --- si el elemento es duplicado y no está en la lista de duplicados, lo añadimos a la lista de duplicados y lo eliminamos de la lista original ---
                duplicateList.append(originalList[j])
                # --- eliminamos el elemento duplicado de la lista original ---
                originalList.pop(j)
            else:
                # --- si el elemento no es duplicado, pasamos al siguiente elemento ---
                j += 1
        i += 1

# --- mostrar la lista de duplicados y la lista sin duplicados ---
print("Lista de duplicados:", duplicateList)
print("Lista sin duplicados:", originalList)
```

2. Escribe un programa en Python para unir dos listas y ordenarlas en orden ascendente.

```python
# --- crear las listas ---
list1 = []
listLen1 = int(input("Teclee la longitud de la lista 1: "))
for i in range(0, listLen1):
    list1.append(input("Elemento {} de la lista 1: ".format(i)))

list2 = []
listLen2 = int(input("Teclee la longitud de la lista 2: "))
for i in range(0, listLen2):
    list2.append(input("Elemento {} de la lista 2: ".format(i)))

# --- unir las listas ---
combinedList = list1 + list2
print("Lista combinada:", combinedList)

# --- ordenar la lista combinada ---
combinedList.sort()
print("Lista combinada ordenada ascendentemente:", combinedList)
```

```python
# --- crear la lista ---
list1 = []
listLen = int(input("Teclee la longitud de la lista: "))
for i in range(0, listLen):
    list1.append(input("Elemento {} de la lista: ".format(i)))

# --- ordenar la lista ---
list1.sort(reverse=True)
print("Lista ordenada descendentemente:", list1)

# --- segundo elemento más grande de la lista ---
if len(list1) < 2:
    print("La lista no tiene suficientes elementos para encontrar el segundo elemento más grande.")
else:
    firstLargest = list1[0]
    i = 1
    while i < len(list1) and list1[i] == firstLargest:
        i += 1
    secondLargest = list1[i]
    print("El segundo elemento más grande de la lista es:", secondLargest)
```

4. Crea un script que cuente el número de elementos más grandes que un determinado número
dado por el usuario (supón una lista numérica).

```python
# --- crear la lista ---
list1 = []
listLen = int(input("Teclee la longitud de la lista: "))
for i in range(0, listLen):
    list1.append(input("Elemento {} de la lista: ".format(i)))

# --- pedir número para comparar
number = int(input("Teclee el número para comparar"))

# --- recorrer la lista y contar los mayores que el número
countGreater = 0
for item in list1:
    if int(item) > number:
        countGreater +=1

# --- imprimir resultado
print("Hay ", countGreater, " números mayoores que ", number, " en la lista ", list1)
```

5. Crea un script dado un número introducido por el usuario o determinado al inicio del
programa, realice la suma de aquellos números que sean divisibles por este.

```python
# --- crear la lista ---
list1 = []
listLen = int(input("Teclee la longitud de la lista: "))
for i in range(0, listLen):
    list1.append(input("Elemento {} de la lista: ".format(i)))

# --- pedir número para hallar divisibles por el
number = int(input("Teclee el número para hallar divisibles por el"))

# --- recorrer la lista agregar divisibles por el número a otra lista
listDivisibles = []
for item in list1:
    if item.isnumeric() and int(item) % number == 0:
        listDivisibles.append(item)

# --- imprimir lista
print("Lista: ", list1)

# --- imprimir resultado
print("Los números divisibles por ", number, " son ", listDivisibles)
```
